```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Orginal image

image = cv2.imread('iris-1.jpg')

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
# 1. Padding: make boarder around the image

replicate = cv2.copyMakeBorder(image, 100, 100, 100, 100, cv2.BORDER_REFLECT)

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(replicate, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()
```

```python
# 2. Cropping: crop the image to a specific region

cropped = image[400:250, 500:250]

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()
```

```python
# 3. Resize: resize the image to a specific size

resized = cv2.resize(image, (200, 200))

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()
```

```python
# 4. Manual copy: copy the image

empty_picture_array = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

np.copyto(empty_picture_array, image)

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(empty_picture_array, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()

```

```python
# 5. Color to Greyscale

grayscale = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(grayscale, cmap='gray'), plt.axis('off')
plt.show()
```

```python
# 6. Color to HSV

hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(hsv), plt.axis('off')
plt.show()  
```

```python
# 7. Color shift

empty_picture_array = np.zeros((image.shape[0], image.shape[1], 3), dtype=np.uint8)

np.copyto(empty_picture_array, image)

empty_picture_array[:, :, 0] += 50  # Set the blue channel to 0
empty_picture_array[:, 0, :] += 50  # Set the green channel to 0
empty_picture_array[0, :, :] += 50  # Set the red channel to 0

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(empty_picture_array, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()
```

```python
# 8. Smoothing: apply a blur to the image

blurred = cv2.GaussianBlur(image, (15, 15), 0)

plt.subplot(1, 2, 1), plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.subplot(1, 2, 2), plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB)), plt.axis('off')
plt.show()  
```
